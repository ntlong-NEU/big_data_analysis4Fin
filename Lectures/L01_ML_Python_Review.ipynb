## Introduction to Machine Learning

### What is Machine Learning?

Machine Learning (ML) is a subfield of artificial intelligence (AI) that focuses on using data and algorithms to imitate the way that humans learn, gradually improving its accuracy.

Simply put, instead of explicitly programming the rules to solve a problem, we "teach" the computer to learn those rules from data.


### Main types of machine learning:

* **Supervised Learning:** The input data is labeled. The model learns to map from the input (**features**) to the output (**labels**). This is the most common type.
    * *Example:* Predicting house prices (input: area, number of bedrooms; output: house price), classifying emails as spam or not (input: email content; output: label "spam" / "not spam").
* **Unsupervised Learning:** The input data is unlabeled. The model automatically finds structures, patterns, or **clusters** in the data.
    * *Example:* Clustering customers based on shopping behavior, dimensionality reduction.
* **Reinforcement Learning:** The model learns through **trial and error**. It interacts with an environment and receives **rewards** or **penalties** for each action.
    * *Example:* Training a game-playing bot, self-driving cars.

### Basic Models and Workflow

In this lecture, we will focus on **Supervised Learning** with two main types of problems:

* **Regression Problem:** Predicting a continuous value. Example: predicting temperature, stock prices.
* **Classification Problem:** Predicting a discrete label/category. Example: classifying cats/dogs, diagnosing diseases.



### Models covered in this lecture

| Model Name | Applicable Problem | Short Description |
| :--- | :--- | :--- |
| **Linear Regression** | Regression | Finds a "straight line" that best fits the data to predict a continuous value. |
| **Logistic Regression** | Classification | Still uses a line (boundary) to divide data, but to classify them into different groups based on probabilities. |
| **K-Nearest Neighbors (KNN)** | Regression & Classification | Predicts based on the nearest "neighbors". For classification, it takes the majority; for regression, it takes the average value. |
| **Decision Tree** | Regression & Classification | Builds a tree-like model of "if-then" rules to make decisions. |
| **Random Forest** | Regression & Classification | Like a vote of multiple Decision Trees (a "forest"). The final result is decided by the majority to increase accuracy. |


### Train-Test Split Rule

In machine learning, we never use the entire dataset to train the model. Instead, we split the dataset into two parts:
- **Training Set:** Accounts for the majority of the data (usually 70-80%), used to "teach" the model.
- **Test Set:** The remaining part (20-30%), used to evaluate the performance of the model after it has been trained. The test set is like a "final exam" that the model has never seen before.

**Why do we do this?** To test the model's **generalization** ability. If a model performs well on the training set but poorly on the test set, it is experiencing **overfitting** or rote learning.


### Model Evaluation Metrics
1. **Regression Models**
**Mean Squared Error (MSE):** The average of the squared errors. It heavily "penalizes" large prediction errors.

$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_{\text{actual}} - y_{\text{predicted}})^2$



**Root Mean Squared Error (RMSE):** The square root of MSE, bringing the error unit back to the same unit as the target variable, making it easier to interpret.

$RMSE = \sqrt{MSE}$

2. **Classification Models**

**Accuracy:** The percentage of data points predicted correctly out of the total number of points.

$Accuracy = \frac{\text{Number of correct predictions}}{\text{Total number of predictions}}$


### Overfitting and Underfitting

After understanding why we need to split data for testing, we will dive deeper into the two most common problems this helps detect.

* **Underfitting:**
    * **Phenomenon:** The model is too simple, failing to capture the underlying patterns of the data. It performs poorly on both the training and test sets.
    * **Example:** Using a straight line to describe a parabolic relationship.
    * **Solution:** Try a more complex model, add features to the data.

* **Overfitting:**
    * **Phenomenon:** The model is too complex, memorizing even the **noise** in the training set. It performs extremely well on the training set but very poorly on the test set.
    * **Example:** Using a wavy curve that passes exactly through all data points in the training set.
    * **Solution:** Use more data, reduce the complexity of the model.



### Standard Workflow of a Machine Learning Project
A typical machine learning project is not just about model training, but an iterative process with multiple stages.
1.  **Data Collection:** Finding and collecting raw data from various sources.
2.  **Data Preprocessing & Exploratory Data Analysis (EDA):** The most critical stage. It includes data cleaning (handling missing values, **outliers**), normalization, and exploring the data to find initial useful insights.
3.  **Model Selection & Training:** Based on the problem and data, selecting a few potential models and training them on the training set.
4.  **Model Evaluation:** Using the test set to evaluate the performance of the models, choosing the best model.
5.  **Parameter Tuning:** Optimizing the parameters of the best model to improve performance.
6.  **Deployment:** Putting the model into a real-world environment so it can make predictions on new data.

## How models learn from data: Manual Calculation and Practice with Scikit-learn

In this section, we will dive deep into each model. For each model, we will learn how it learns through a manual calculation example, followed immediately by practice with the Scikit-learn library to compare results.

**Note:** The datasets used in this section are very small, primarily intended to *illustrate how the model works* and how to calculate evaluation metrics. Therefore, the results and scores of the models may not accurately reflect their actual "learning" quality.

For practice, we will use **Scikit-learn**, the most popular machine learning library in Python. Scikit-learn's strength is its consistent and easy-to-use programming interface. The general workflow usually involves 3 steps:

```python
# Initialize common libraries and data for Part 3
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.neighbors import KNeighborsRegressor, KNeighborsClassifier
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_squared_error, accuracy_score, roc_curve, roc_auc_score

# Regression Data
X_reg = np.array([[1], [1.5], [2.5], [3], [4], [5], [5.5], [6], [7], [8]])
y_reg = np.array([15, 18, 22, 25, 28, 35, 36, 40, 45, 52])
X_train_reg, X_test_reg = X_reg[:8], X_reg[8:]
y_train_reg, y_test_reg = y_reg[:8], y_reg[8:]

# Classification Data
X_cls = np.array([[1], [2], [3], [4.5], [5.5], [6], [7], [8], [4], [9]])
y_cls = np.array([0, 0, 0, 0, 1, 1, 1, 1, 0, 1]) # 0: Fail, 1: Pass
X_train_cls, X_test_cls = X_cls[:8], X_cls[8:]
y_train_cls, y_test_cls = y_cls[:8], y_cls[8:]
```

### Regression problem

+ Problem and data

**Problem:** Predicting salary (million VND) based on years of experience.

**Data (10 observations):**
| Experience (years) | Salary (million) |
|:---:|:---:|
| 1 | 15 |
| 1.5 | 18 |
| 2.5 | 22 |
| 3 | 25 |

**Split Rule:**
- **Train Set:** First 8 data points.

+ Linear Regression
**How does the model learn?**

The Linear Regression model tries to find a line equation `y_predicted = β₀ + β₁ * x` that best fits the **train set data**. Where:
- `β₁` is the **slope**: Indicates how much the Salary changes when Experience increases by 1 year.
- `β₀` is the **intercept**: Is the base salary when experience is 0.
It "learns" by continuously adjusting `β₀` and `β₁` so that the mean squared error (MSE) between the predicted and actual values on the **train set** is minimized.


After training, the model learned the coefficients as **`β₀ ≈ 10.19`** and **`β₁ ≈ 4.82`**. We will use these coefficients for calculations.
**Salary ≈ 10.19 + 4.82 * Experience**

**Prediction and Evaluation:**
- With Experience = 7: `y_predicted` = 10.19 + 4.82 * 7 ≈ **43.95** (Actual: 45)
- With Experience = 8: `y_predicted` = 10.19 + 4.82 * 8 ≈ **48.78** (Actual: 52)

**Calculating RMSE:**
$RMSE = \sqrt{\frac{(45 - 43.95)^2 + (52 - 48.78)^2}{2}} \approx 2.395$

+ Practice with Scikit-learn

```python
model_lr = LinearRegression()
model_lr.fit(X_train_reg, y_train_reg)
preds_lr = model_lr.predict(X_test_reg)
rmse_lr = np.sqrt(mean_squared_error(y_test_reg, preds_lr))
print(f"Hệ số β₁ (slope): {model_lr.coef_[0]:.4f}")
print(f"Hệ số β₀ (intercept): {model_lr.intercept_:.4f}")
print(f"Dự đoán: {preds_lr.flatten()}")
print(f"RMSE: {rmse_lr:.4f}")
```

### K-Nearest Neighbors (KNN)
**How does the model learn?**

KNN is a **"lazy learner"**. It does not actually learn any function. Instead, it memorizes the entire **train set data**. When it needs to predict for a new data point (from the test set), it will:
1. Calculate the distance from this new point to all points in the train set.
2. Find the K points in the train set that have the closest distance (K nearest "neighbors").
3. Take the average value (salary) of those K neighbors as the prediction result.


Consider **K=3**.

**Prediction and Evaluation:**
- With Experience = 7 (neighbors: 6, 5.5, 5): `y_predicted` = (40 + 36 + 35) / 3 = **37** (Actual: 45)
- With Experience = 8 (neighbors: 6, 5.5, 5): `y_predicted` = (40 + 36 + 35) / 3 = **37** (Actual: 52)

**Calculating RMSE:**
$RMSE = \sqrt{\frac{(45 - 37)^2 + (52 - 37)^2}{2}} \approx 12.02$

### Practice with Scikit-learn

```python
model_knn_reg = KNeighborsRegressor(n_neighbors=3)
model_knn_reg.fit(X_train_reg, y_train_reg)
preds_knn_reg = model_knn_reg.predict(X_test_reg)
rmse_knn_reg = np.sqrt(mean_squared_error(y_test_reg, preds_knn_reg))
print(f"Dự đoán: {preds_knn_reg.flatten()}")
print(f"RMSE: {rmse_knn_reg:.4f}")
```

```python
# Visualize
plt.figure(figsize=(10, 6))
plt.scatter(X_train_reg, y_train_reg, color='blue', label='Dữ liệu Train')
plt.scatter(X_test_reg, y_test_reg, color='green', label='Dữ liệu Test (Thực tế)')
plt.scatter(X_test_reg, preds_knn_reg, color='red', marker='x', s=100, label='Dữ liệu Test (Dự đoán)')

# Draw KNN prediction line
X_plot = np.linspace(0, 10, 100).reshape(-1, 1)
y_plot = model_knn_reg.predict(X_plot)
plt.plot(X_plot, y_plot, color='red', linestyle='--', label='Đường dự đoán KNN (k=3)')

plt.title('KNN Regressor - Trực quan hóa')
plt.xlabel('Kinh nghiệm (năm)')
plt.ylabel('Lương (triệu)')
plt.legend()
plt.grid(True)
plt.savefig('ml_image2.png')
```

### Decision Tree
**How does the model learn?**

Imagine the Decision Tree model learns by asking a series of "Yes/No" questions to split the **training data**. For example: `Experience < 3.5 years?`. At each step, it will find a question (based on a feature and a threshold) that helps split the data into two subgroups. The criterion for choosing the best question is to find a split such that the **Mean Squared Error (MSE)** of the subgroups is minimized. This ensures that the values in each subgroup are as close to each other as possible. This process repeats until a stopping condition is met. The result is a tree of rules. When predicting, a new data point will go from the root down the branches according to the rules and receive a predicted value which is the average of the points in the final leaf it falls into.


**Initial calculation (root node):**
Before splitting, the model needs a reference point. It will calculate the MSE for the entire training set. This is the initial error level if we just predict using the mean value of the entire dataset.
- Mean of salary in the training set: `(15+18+22+25+28+35+36+40) / 8 = 27.375`
- Initial MSE: `((15-27.375)² + ... + (40-27.375)²) / 8 ≈ 71.23`
The goal of finding a split threshold is to reduce this metric as low as possible.

**Finding the optimal split threshold (for the root node):**
The model will try all possible split thresholds (the midpoint of two consecutive experience values) and choose the threshold that gives the lowest **MSE after split**. The calculation formula is: 

$MSE_{\text{after split}} = \frac{n_{\text{left}}}{n_{\text{total}}} MSE_{\text{left}} + \frac{n_{\text{right}}}{n_{\text{total}}} MSE_{\text{right}}$

| Split threshold (<) | MSE (Left) | MSE (Right) | MSE after split |
|:---:|:---:|:---:|:---:|
| 1.25 | 0.0 | 55.27 | 48.36 |
| 2.0 | 2.25 | 33.56 | 25.73 |
| 2.75 | 8.22 | 26.24 | 19.48 |
| 3.5 | 14.5 | 18.69 | 16.60 |
| **4.5** | **21.84** | **4.67** | **15.40** |
| 5.25 | 41.81 | 4.0 | 32.35 |
| 5.75 | 47.63 | 0.0 | 41.68 |

-> The model chooses `Experience < 4.5` as the first split rule because it gives the lowest MSE after split (reduced from 71.23 to 15.40). The predicted value for the branch `Experience >= 4.5` will be the mean of the points in that branch: (35+36+40)/3 = 37.


**Prediction and Evaluation:**
- With Experience = 7 (>= 4.5): `y_predict` = **37** (Actual: 45)
- With Experience = 8 (>= 4.5): `y_predict` = **37** (Actual: 52)

**Calculating RMSE:**
$RMSE = \sqrt{\frac{(45 - 37)^2 + (52 - 37)^2}{2}} \approx 12.02$

### Random Forest
**How does the model learn?**

Random Forest builds many different Decision Trees and combines their predictions. It learns by:
1. Creating multiple subsamples of data from the **training set** (by random sampling with replacement).
2. Training a separate Decision Tree on each of those data subsamples. To increase diversity, each tree is only allowed to consider a random subset of features at each split.
3. When predicting, the model will take the average result from all the trees in the "forest".



**Prediction and Evaluation:**
The actual prediction result from the code is the average of 3 trees, yielding `y_predict` as **37.0** for both test points. (With `random_state=42`, the result will always be 37.0).
- With Experience = 7: `y_predict` = **37.0** (Actual: 45)
- With Experience = 8: `y_predict` = **37.0** (Actual: 52)

**Calculating RMSE:**
$RMSE = \sqrt{\frac{(45 - 37)^2 + (52 - 37)^2}{2}} \approx 12.02$

## Classification problem

**Problem:** Classify a student as "Pass" or "Fail" in a subject based on "Self-study hours" per week.

**Data (10 observations):**
| Self-study hours | Result |
|:---:|:---:|
| 1 | Fail |
| 2 | Fail |
| 3 | Fail |
| 4.5 | Fail |
| 5.5 | Pass |
| 6 | Pass |
| 7 | Pass |
| 8 | Pass |
| 4 | Fail |
| 9 | Pass |

**Splitting rule:**
- **Train Set:** First 8 data points.
- **Test Set:** Last 2 data points `(4, Fail)` and `(9, Pass)`.

### Logistic Regression
**How does the model learn?**
The Logistic Regression model learns to transform an input value `x` into a probability (from 0 to 1) using the **Sigmoid function**. This probability is the `y_predict` for class 1 ('Pass').

After training on the train set, the model learned the optimal coefficients which are **`β₀ ≈ -5.5288`** and **`β₁ ≈ 1.1590`**. So the probability prediction formula is:

$y_{\text{predict}} = P(y=1|x) = \frac{1}{1 + e^{-(-5.5288 + 1.1590 * x)}}$

If this probability > 0.5, the model will classify as 'Pass', otherwise it is 'Fail'.



**Prediction and Evaluation:**
- With Hours = 4: $P(y=1|x=4) = \frac{1}{1 + e^{-(-5.5288 + 1.1590 * 4)}} = \frac{1}{1 + e^{0.8928}} \approx 0.29$ Since 0.29 < 0.5, the prediction is **Fail**. (Correct)
- With Hours = 9: $P(y=1|x=9) = \frac{1}{1 + e^{-(-5.5288 + 1.1590 * 9)}} = \frac{1}{1 + e^{-4.9022}} \approx 0.99$ Since 0.99 > 0.5, the prediction is **Pass**. (Correct)

**Calculating Accuracy:**
$Accuracy = \frac{2 \text{ (correct predictions)}}{2 \text{ (total)}} = 100\%$

#### Practice with Scikit-learn

```python
model_log_reg = LogisticRegression(random_state=42)
model_log_reg.fit(X_train_cls, y_train_cls)
preds_log_reg = model_log_reg.predict(X_test_cls)
acc_log_reg = accuracy_score(y_test_cls, preds_log_reg)
print(f"Coefficient β₁ (coef): {model_log_reg.coef_[0][0]:.4f}")
print(f"Coefficient β₀ (intercept): {model_log_reg.intercept_[0]:.4f}")
print(f"Prediction: {preds_log_reg}")
print(f"Accuracy: {acc_log_reg:.2f}")
```

```python

# Visualize
plt.figure(figsize=(10, 6))
# Train data
plt.scatter(X_train_cls[y_train_cls==0], y_train_cls[y_train_cls==0], color='blue', label='Train - Fail')
plt.scatter(X_train_cls[y_train_cls==1], y_train_cls[y_train_cls==1], color='orange', label='Train - Pass')
# Test data
plt.scatter(X_test_cls[y_test_cls==0], y_test_cls[y_test_cls==0], color='blue', marker='s', s=100, label='Test - Fail (Actual)')
plt.scatter(X_test_cls[y_test_cls==1], y_test_cls[y_test_cls==1], color='orange', marker='s', s=100, label='Test - Pass (Actual)')

# Plot sigmoid line and decision boundary
X_plot = np.linspace(0, 10, 300).reshape(-1, 1)
y_plot_proba = model_log_reg.predict_proba(X_plot)[:, 1]
plt.plot(X_plot, y_plot_proba, color='red', label='Probability of "Pass" (Sigmoid)')
# Decision boundary is where probability = 0.5
decision_boundary = -model_log_reg.intercept_ / model_log_reg.coef_[0]
plt.axvline(x=decision_boundary, color='green', linestyle='--', label='Decision boundary')
plt.axhline(y=0.5, color='grey', linestyle=':')

plt.title('Logistic Regression - Visualization')
plt.xlabel('Self-study hours')
plt.ylabel('Result (0: Fail, 1: Pass) / Probability')
plt.legend()
plt.grid(True)
plt.savefig('ml_image8.png')
```

### K-Nearest Neighbors (KNN)
**How does the model learn?**
Similar to KNN regression, the model memorizes the entire **training set**. When predicting for a new data point, it will:
1. Calculate the distance from this new point to all points in the training set.
2. Find the K points in the training set with the closest distance.
3. Let these K neighbors "vote": the label that appears most often among the neighbors will be the predicted result.



**Prediction and Evaluation (K=3):**
- With Hours = 4 (neighbors: 4.5, 3, 5.5): 2 'Fail', 1 'Pass' -> `y_predict` = **Fail**. (Correct)
- With Hours = 9 (neighbors: 8, 7, 6): 3 'Pass' -> `y_predict` = **Pass**. (Correct)

**Calculating Accuracy:**
$Accuracy = \frac{2}{2} = 100\%$

### Decision Tree
**How does the model learn?**

From the **training data**, the model learns by continuously asking "Yes/No" questions to split the data. It iterates through all features and possible split thresholds to find the best question. "Best" here means the question that creates the two "purest" subgroups. Purity is measured by the **Gini index**. The closer Gini is to 0, the purer the group. This process repeats, creating a tree structure. When making a prediction, a new data point follows the questions on the tree to reach a leaf node, and is assigned the label of the majority class in that leaf node.

**Gini Formula:** Where C is the number of classes, $p_i$ is the proportion of class i in a node.

$Gini = 1 - \sum_{i=1}^{C} p_i^2$



**Initial calculation (root node):**
Similar to the regression problem, the model also needs a reference point. It will calculate the Gini index for the entire training set to measure the initial "impurity".
- In the 8 points of the train set, there are 4 'Fail' points (class 0) and 4 'Pass' points (class 1).
- Proportion of class 0: p₀ = 4/8 = 0.5
- Proportion of class 1: p₁ = 4/8 = 0.5
- Initial Gini: $Gini = 1 - (0.5^2 + 0.5^2) = 1 - (0.25 + 0.25) = 0.5$
The goal of finding a split threshold is to reduce this Gini index as low as possible.

**Finding the optimal split threshold (for the root node):**
The model will try different split thresholds and choose the one that has the lowest **Gini after split**. The formula is:

$Gini_{\text{after split}} = \frac{n_{\text{left}}}{n_{\text{total}}}Gini_{\text{left}} + \frac{n_{\text{right}}}{n_{\text{total}}}Gini_{\text{right}}$

| Split threshold (<) | Gini (Left) | Gini (Right) | Gini after split |
|:---:|:---:|:---:|:---:|
| 1.5 | 0.0 | 0.490 | 0.428 |
| 2.5 | 0.0 | 0.444 | 0.333 |
| 3.75 | 0.0 | 0.320 | 0.200 |
| **5.0** | **0.0** | **0.0** | **0.0** |
| 5.75 | 0.320 | 0.0 | 0.200 |
| 6.5 | 0.444 | 0.0 | 0.333 |
| 7.5 | 0.490 | 0.0 | 0.428 |

-> The model chooses `Study hours < 5.0` as the optimal split rule because it gives the lowest Gini after split (reduced from 0.5 to 0.0), creating two completely pure subgroups.


**Prediction and Evaluation:**
- With Hours = 4 (< 5.0): `y_predicted` = **Fail**. (Correct)
- With Hours = 9 (>= 5.0): `y_predicted` = **Pass**. (Correct)

**Calculate Accuracy:**
$Accuracy = \frac{2}{2} = 100\%$

### Further discussion
**When do algorithms stop learning?**

The "learning" process does not happen forever. Models have different stopping conditions:
* **Iterative models (Linear/Logistic Regression):** Usually stop after a predefined number of iterations (`max_iter`), or when the level of improvement (e.g., the change in the loss function) between iterations becomes very small and insignificant.
* **Decision Tree:** Stops when a branch (node) cannot be split any further. This happens when:
    * All data points in that node belong to the same class (a "pure" node).
    * The maximum depth of the tree (`max_depth`) allowed by the user has been reached.
    * The number of samples in a node is too small to be split further (`min_samples_split`).
* **KNN:** As a "lazy learning model", it does not have an actual training process, so there is no learning stopping condition. It simply stores the entire data.


### When to use Data Scaling?
**Data scaling** is the process of bringing data columns (features) to the same **scale**. This is a critically important step.

* **Why is it necessary?** Imagine you have 2 features: `experience` (from 1-10) and `previous month salary` (from 15,000,000 - 50,000,000). For **distance-based** algorithms like **KNN**, the absolute difference in the value of the `salary` feature will completely "overpower" the `experience` feature. The model will almost exclusively rely on salary to calculate distance, ignoring important information from experience.

* **Which models need it?**
    * **NEEDED:** Models sensitive to the scale of the data, including distance-based models (**KNN**) and models optimized by **Gradient Descent** such as **Linear Regression, Logistic Regression**.
    * **NOT NEEDED:** Rule-based models like **Decision Tree** and **Random Forest**. These models consider each feature independently at each split, so they are not affected by differences in scale between features.

### K-Means Clustering
K-Means is a popular clustering algorithm that partitions data into **K** distinct clusters based on feature similarity.

**How it works (Intuition):**
1. **Initialization:** Randomly select K points as initial cluster centers (centroids).
2. **Assignment:** Assign each data point to the nearest centroid (using Euclidean distance).
3. **Update:** Recalculate the centroids as the mean of all points assigned to that cluster.
4. **Repeat:** Repeat steps 2 and 3 until the centroids no longer change significantly.

> [!NOTE]
> The number of clusters **K** must be specified in advance. A common method to find the optimal K is the **Elbow Method**.

#### Practice with Scikit-learn

```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

# Generate dummy data with 3 distinct clusters
X_unsup, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=42)

# Initialize and fit K-Means
kmeans = KMeans(n_clusters=3, random_state=42, n_init='auto')
kmeans.fit(X_unsup)

# Predict clusters
y_kmeans = kmeans.predict(X_unsup)

# Visualize the clusters and centroids
plt.figure(figsize=(8, 5))
plt.scatter(X_unsup[:, 0], X_unsup[:, 1], c=y_kmeans, s=50, cmap='viridis', alpha=0.6)
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, marker='X', label='Centroids')
plt.title('K-Means Clustering Example')
plt.legend()
plt.show()
```

## Advanced evaluation metrics

Basic metrics like RMSE and Accuracy are very useful, but they do not always tell the whole story.

### For regression problems
**Mean Absolute Error (MAE):** The average of the absolute errors. This metric is less affected by outliers than MSE.

$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_{\text{actual}} - y_{\text{predicted}}|$

**Coefficient of determination R² (R-squared):** Tells us how good the model is compared to a simple model (e.g., always predicting the mean). It indicates what percentage of the variance in the data is explained by the model. The closer R² is to 1, the better the model.

$R^2 = 1 - \frac{\sum_{i=1}^{n} (y_{\text{actual}, i} - y_{\text{predicted}, i})^2}{\sum_{i=1}^{n} (y_{\text{actual}, i} - \bar{y}_{\text{actual}})^2}$

Where $\bar{y}_{\text{actual}}$ is the mean value of all actual data points.


### For classification problems
**The problem with Accuracy:** Imagine a rare disease prediction problem, where 99% of people are "normal" and 1% are "sick". A naive model that always predicts "normal" would have an accuracy of 99%! However, this model is completely useless because it fails to detect any sick cases.

This is where we need other metrics:

* **Confusion Matrix:** A table that details correct and incorrect predictions.
    * **True Positive (TP):** Predicted 'Sick' and actually 'Sick'. (Good)
    * **True Negative (TN):** Predicted 'Normal' and actually 'Normal'. (Good)
    * **False Positive (FP):** Predicted 'Sick' but actually 'Normal'. (False alarm)
    * **False Negative (FN):** Predicted 'Normal' but actually 'Sick'. (Missed, very dangerous!)



* **Precision:** Out of all the 'Sick' predictions, how many were correct? Helps reduce false alarms.

$Precision = \frac{TP}{TP + FP}$

* **Recall (Sensitivity):** Out of all the actually 'Sick' people, how many did the model detect? Helps reduce missed cases.

$Recall = \frac{TP}{TP + FN}$

* **F1-Score:** The harmonic mean of Precision and Recall, useful when both need to be balanced.

* **ROC Curve and AUC Metric:**
    * **ROC Curve (Receiver Operating Characteristic):** A graph showing the classification ability of a model at all different **thresholds**. It plots the **True Positive Rate (Recall)** against the **False Positive Rate**.
    * **Area Under the Curve (AUC):** The area under the ROC curve. AUC measures the overall two-dimensional performance of the model. The closer AUC is to 1, the better the model. A randomly guessing model will have an AUC = 0.5.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

clf_models = {
    "Logistic Regression": LogisticRegression(random_state=42),
    "KNeighbors Classifier": KNeighborsClassifier(n_neighbors=5),
    "Decision Tree": DecisionTreeClassifier(max_depth=3, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=10, random_state=42)
}

fig, axes = plt.subplots(1, 4, figsize=(24, 5)) 

clf_results = []

# Use enumerate to get both index (i) and item in the loop
for i, (name, model) in enumerate(clf_models.items()):
    model.fit(X_train_c_scaled, y_train_c)
    y_pred_c = model.predict(X_test_c_scaled)
    
    acc = accuracy_score(y_test_c, y_pred_c)
    prec = precision_score(y_test_c, y_pred_c, pos_label=0) 
    rec = recall_score(y_test_c, y_pred_c, pos_label=0)
    
    clf_results.append({'Model': name, 'Accuracy': acc, 'Precision': prec, 'Recall': rec})
    
    print(f"--- {name} ---")
    print(f"Accuracy: {acc:.4f}, Precision: {prec:.4f}, Recall: {rec:.4f}")
    
    # 2. Plot Confusion Matrix on the corresponding subplot (axes[i])
    cm = confusion_matrix(y_test_c, y_pred_c)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=cancer.target_names, yticklabels=cancer.target_names,
                ax=axes[i]) # <--- Specify the subplot to plot on
    
    axes[i].set_title(f'Confusion Matrix - {name}')
    axes[i].set_xlabel('Predicted')
    axes[i].set_ylabel('Actual')

# 3. Adjust layout to look nice and show/save figure
plt.tight_layout() # Automatically adjust spacing between subplots
plt.savefig('all_confusion_matrices.png') # Save the figure containing all matrices
plt.show() # Show figure
```
